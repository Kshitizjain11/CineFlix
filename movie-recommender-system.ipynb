##### Original language is also a good column for recommendation but since the data is very imbalanced more skewed towards en (english) movies. We may drop this. 

```python
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.head()
```

```python
# For data precprocessing removing null and duplicate data, if any
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.isnull().sum()
```

```python
movies.duplicated().sum()
```

```python
movies.iloc[0].genres
```

```python
# converting the given list of Dict into list of strings ['Action','Fantasy','Adventure'...]
from ast import literal_eval

def convert(obj):
    array = []
    for i in literal_eval(obj):
        array.append(i['name'])
    
    return array
```

```python
convert(movies.iloc[0].genres)
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
# Similarly, doing the same for keywords (having same key -> name)

movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.iloc[0]
```

```python
# Creating a fn for cast and crew
def convert_for_cast(obj):
    counter = 0
    array = []
    for i in literal_eval(obj):
        if counter != 3:
            array.append(i['name'])
            counter += 1
        else :
            break
    return array
```

```python
convert_for_cast(movies.iloc[0].cast)
```

```python
movies['cast'] = movies['cast'].apply(convert_for_cast)
```

```python
movies.iloc[0]
```

```python
literal_eval(movies.crew[0])
# since for a movie one of the most important aspect (apart from cast) is the director and we filter out the director name 
```

```python
# Creating a fn for crew (director)
def fetch_director(obj):
    array = []
    for i in literal_eval(obj):
        if i['job'] == 'Director':
            array.append(i['name'])
            break
    return array
fetch_director(movies.iloc[0].crew)
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head(1)
```

```python
# converting the overviw (str) into list of strings so later we could concatenate with others

movies['overview'] = movies['overview'].apply(lambda x : x.split())
```

```python
movies.head()
```

```python
# removing spaces in each column
ls = ["genres",'keywords','cast','crew'] 
for element in ls:
    movies[element] = movies[element].apply(lambda x : [i.replace(" ",'') for i in x]) 
```

```python
movies.head()
```

```python
# creating a new column tags and merging the 5 columns

movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.head()
```

```python
new_df = movies[['movie_id','title','tags']]
```

```python
new_df.head()
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x : " ".join(x))
```

```python
new_df.tags[0]
```

```python
# it is recommended to convert the string into lowercase.

new_df['tags'] = new_df.tags.apply(lambda x : x.lower())
new_df.tags[0]
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
stem('in the 22nd century, a paraplegic marine is dispatched to the moon pandora on a unique mission, but becomes torn between following orders and protecting an alien civilization. action adventure fantasy sciencefiction cultureclash future spacewar spacecolony society spacetravel futuristic romance space alien tribe alienplanet cgi marine soldier battle loveaffair antiwar powerrelations mindandsoul 3d samworthington zoesaldana sigourneyweaver jamescameron')
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
# vectorizing the tags and removing stop words
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors
```

```python
# for i in cv.get_feature_names_out():
#     print(i)

cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
print(similarity.shape)
similarity
```

```python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title']==movie].index[0]
    distances = similarity[movie_index]
    movie_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movie_list:
        print(new_df.iloc[i[0]].title)

    
```

```python
recommend('Batman Begins')
```
